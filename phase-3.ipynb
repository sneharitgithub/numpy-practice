# Advanced Numpy Operations with Business Examples

## Boolean Masking - Sales Filtering


```python

import numpy as np
```

### Show me all days where sales were more than ₹5,000.

```python
# Business Scenario

sales = np.array([3200, 4500, 7000, 1200, 9800, 5400])

high_sales = sales[sales > 5000]
print("High sales days:", high_sales)
```

## Conditional Replacement – Discount Logic

### Give 10% bonus points to customers who spent more than ₹5,000.

```python
spending = np.array([3000, 6000, 4500, 8000])

bonus_points = np.where(spending > 5000, spending * 0.10, 0)
print("Bonus points:", bonus_points)
```

## Broadcasting – Price Increase Across Products

### Company increases GST by ₹5 on all product prices.

```python
prices = np.array([100, 200, 300, 400])
updated_prices = prices + 5
print(updated_prices)
```

## Aggregation with Axis – Regional Sales Report

### Sales of 3 products across 3 regions.

```python
sales = np.array([
    [100, 200, 300],   # North
    [150, 250, 350],   # South
    [200, 300, 400]    # West
])

print("Total sales per product:", np.sum(sales, axis=0))
print("Total sales per region:", np.sum(sales, axis=1))
```

## Sorting & Ranking – Top Performing Stores

### Find top performing store by revenue.

```python
revenue = np.array([120000, 90000, 150000, 110000])

sorted_revenue = np.sort(revenue)[::-1]
print("Top revenues:", sorted_revenue)
```

## Fancy Indexing – Selected Branch Analysis

### Sales data of branch 1, 3 and 5 only.

```python
sales = np.array([5000, 7000, 6500, 8000, 7200])
selected_branches = sales[[0, 2, 4]]

print(selected_branches)
```

## Handling Missing Data – Incomplete Reports

### Some stores didn’t report sales (missing values).

```python
sales = np.array([5000, np.nan, 7000, np.nan, 8000])

clean_sales = sales[~np.isnan(sales)]
print("Valid sales:", clean_sales)
print("Average sales:", np.nanmean(sales))
```

## Profit & Loss Calculation – Business KPI

## Calculate profit = revenue − cost.

```python
revenue = np.array([10000, 15000, 20000])
cost = np.array([7000, 9000, 12000])

profit = revenue - cost
print("Profit:", profit)
```

## Percentage Contribution – Revenue Share

### What percentage does each product contribute to total revenue?

```python
revenue = np.array([20000, 30000, 50000])

percentage = (revenue / np.sum(revenue)) * 100
print("Revenue contribution %:", percentage)
```

## Mini Business Case

### Monthly sales data of a company.

```python
sales = np.array([12000, 18000, 25000, 9000, 30000])

print("Total sales:", np.sum(sales))
print("Average sales:", np.mean(sales))
print("High performing months:", sales[sales > 20000])
print("Best month sale:", np.max(sales))
```
